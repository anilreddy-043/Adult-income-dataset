```python
Q1 = df['Capital-Loss'].quantile(0.25)
Q3 = df['Capital-Loss'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Capital-Loss'] < lower) | (df['Capital-Loss'] > upper)]
outliers
```

```python
Q1 = df['Capital-Gain'].quantile(0.25)
Q3 = df['Capital-Gain'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Capital-Gain'] < lower) | (df['Capital-Gain'] > upper)]
outliers
```

```python
Q1 = df['Hours-Per-Week'].quantile(0.25)
Q3 = df['Hours-Per-Week'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Hours-Per-Week'] < lower) | (df['Hours-Per-Week'] > upper)]
outliers
```
